```python
import statistics

age = [22, 21, 20, 20, 21, 21, 21, 21, 20, 20, 20, 22, 21, 20]

print("Mean:", statistics.mean(age))
print("Median:", statistics.median(age))
print("Mode:", statistics.mode(age))
print("Sample variance:", statistics.variance(age))
print("Sample standard deviation:", statistics.stdev(age))
```
